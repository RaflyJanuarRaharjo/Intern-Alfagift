```python
import pandas as pd
import glob
import os

# 1. Hubungkan Google Drive
from google.colab import drive
drive.mount('/content/drive')

# 2. Lokasi folder yang berisi file
folder = '/content/drive/MyDrive/.Master Produk Alfagift'

# 3. Ambil semua file .xls (yang sebenarnya teks/tab-separated)
files = glob.glob(os.path.join(folder, '*.xls'))

print(f"Jumlah file ditemukan: {len(files)}")

# 4. Baca semua file menggunakan read_csv dengan separator TAB
df_list = []

for file in files:
    # sep='\t' karena data dipisahkan oleh tab
    # skiprows=2 karena ada judul 'Report Master Group PLU' dan baris kosong di atas header
    try:
        df = pd.read_csv(file, sep='\t', skiprows=2)
        df_list.append(df)
    except Exception as e:
        print(f"Gagal membaca file {os.path.basename(file)}: {e}")

# 5. Gabungkan semua file
if df_list:
    df_merge = pd.concat(df_list, ignore_index=True)
    # 6. Cek hasil
    print("Jumlah baris total:", len(df_merge))
    print("Jumlah kolom:", len(df_merge.columns))
    display(df_merge.head())
else:
    print("Tidak ada data yang berhasil dibaca.")
```

```python
# Daftar tag yang diinginkan
tags_to_keep = ['D', 'G', 'S', 'W', '6']

# Filter DataFrame berdasarkan kolom TAG
df_filtered = df_merge[df_merge['TAG'].isin(tags_to_keep)]

# Tampilkan hasil
print(f"Jumlah baris setelah difilter: {len(df_filtered)}")
display(df_filtered.head())
```
